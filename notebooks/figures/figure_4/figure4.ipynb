```python
import json
import gzip
from plot_utils import plot_figure4, fig_json_hook
```

```python
FIGURE_OUT_PATH = "../../../data/output/figures/figure_4.png"
DATA_PATH = "figure4_data.json.gz"

'''Restore variables - note this may take a few minutes to decompress and load'''

with gzip.open(DATA_PATH, "rt", encoding="utf-8") as f:
    loaded = json.load(f, object_hook=fig_json_hook)


# Data for 4a-b (ExCALIBR MSH2)
MSH2_scoreset_2018 = loaded["scoreset_2018"]
MSH2_scoreset = loaded["scoreset"]
excalibr_calibration_summary = loaded["indv_summary"]
excalibr_fits = loaded["fits"]
MSH2_score_range = loaded["score_range"]
excalibr_n_c = loaded["n_c"]
MSH2_n_samples = loaded["n_samples"]
MSH2_scoreset_flipped = loaded["scoreset_flipped"]

# Data for 4c (ExCALIBR confusion matrices)
excalibr_oob_evidence = loaded["danzs_oob"]
auth_assignments = loaded["auths_oob"]
datasets_considered = loaded["datasets"]

# Data for 4d (gene-specific predictor calibration overview)
prior = loaded["prior"]
Post_p = loaded["Post_p"]
Post_b = loaded["Post_b"]
p_data = loaded["p_data"]
b_data = loaded["b_data"]

# Data for 4e (REVEL gene-specific vs. genome-wide aggregate comparison)
gene_4e = loaded["gene_4e"]
dist_4e = loaded["dist_4e"]
labdat_4e = loaded["labdat_4e"]
snvdf_4e = loaded["snvdf_4e"]
gene_specific_thresholds_4e = loaded["sorted_thresholds_4e"]
genome_wide_thresholds_4e = loaded["oldsorted_thresholds_4e"]

# Data for 4f (REVEL gene-specific vs. genome-wide heatmap)
dist_4f = loaded["dist_4f"]
finalout_4f = loaded["finalout_4f"]
```

```python
fig = plot_figure4(
    # Data for 4a-b (ExCALIBR MSH2)
    MSH2_scoreset_2018, MSH2_scoreset, excalibr_calibration_summary, excalibr_fits, MSH2_score_range, excalibr_n_c, MSH2_n_samples, MSH2_scoreset_flipped,
    # Data for 4c (ExCALIBR confusion matrices)
    excalibr_oob_evidence, auth_assignments, datasets_considered,
    # Data for 4d (gene-specific predictor calibration overview)
    prior, Post_p, Post_b, p_data, b_data,
    # Data for 4e (REVEL gene-specific vs. genome-wide aggregate comparison)
    gene_4e, dist_4e, labdat_4e, snvdf_4e, gene_specific_thresholds_4e, genome_wide_thresholds_4e,
    # Data for 4f (REVEL gene-specific vs. genome-wide heatmap)
    dist_4f, finalout_4f
)

fig.savefig(FIGURE_OUT_PATH,dpi=300,bbox_inches='tight')
```
