This notebook processes the large pillar project dataframe and extracts data relevant to figure 2. There are three Excel file outputs: 1) a complete file with all data from SGE and VAMP-seq, 2) only SGE data, and 3) only VAMP-seq data

```python
from pathlib import Path
import pandas as pd
```

```python
save = True
bigppj_df = '../../../data/output/maves/integrated_variant_effect_dataset.tsv.gz' #Path to pillar project dataframe

sge_genes = ['BARD1', 'PALB2', 'BRCA2', 'RAD51D', 'XRCC2', 'CTCF', 'SFPQ']
vampseq_genes = ['G6PD', 'TSC2', 'F9']
```

```python
df = pd.read_csv(bigppj_df, sep='\t', low_memory=False) #Reads the dataframe


sge_subset = df.loc[(df['Gene'].isin(sge_genes)) & (df['Dataset'].str.contains('_IGVF'))] #Gets all SGE data
vampseq_subset = df.loc[df['Gene'].isin(vampseq_genes)] #Pulls all VAMP-seq data, including every F9 antibody
                                                          #dataset and both TSC2 domains -- the panel notebooks
                                                          #do their own per-gene filtering/splitting downstream


if save:
    Path("../../../data/intermediate/figures/figure_2").mkdir(parents=True, exist_ok=True)

    print('Saving SGE...')
    sge_subset.to_excel('../../../data/intermediate/figures/figure_2/SGEsubset.xlsx', index = False)

    print('Saving VAMP-seq...')
    vampseq_subset.to_excel('../../../data/intermediate/figures/figure_2/VAMPseqsubset_wDups.xlsx', index = False)


    print('Saving CAVA subset...')
    cava_subset = pd.concat([sge_subset, vampseq_subset])
    cava_subset.to_excel('../../../data/intermediate/figures/figure_2/CAVAseqsubset.xlsx', index = False)

    print('Done!')

else:
    print(f'No files saved')
```
